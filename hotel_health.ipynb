```python
df_clean = df.copy()

# Remove negative ADR
df_clean = df_clean[df_clean["adr"] >= 0]

# Create total guests
df_clean["total_guests"] = (
    df_clean["adults"] +
    df_clean["children"] +
    df_clean["babies"]
)

# Remove bookings with no guests
df_clean = df_clean[df_clean["total_guests"] > 0]

# Remove helper column from final dataset
df_clean = df_clean.drop(columns=["total_guests"])

print("Final shape:", df_clean.shape)
```
